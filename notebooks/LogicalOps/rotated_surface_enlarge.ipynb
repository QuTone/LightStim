# Rotated surface-code enlargement

**[DEMO]** — Uses the packaged `enlarge` logical operation.

Prepare a distance-3 patch, run one source SE round, enlarge to distance 5,
and read out in the preparation basis. `ROUNDS` counts all enlarged-code
SE rounds; only added data qubits are initialized.


```python
import sys
from pathlib import Path

ROOT = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / "lightstim").is_dir())
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from lightstim.protocols import RotatedSurfaceEnlargeExperiment
```

```python
DISTANCE = 3
TARGET_DISTANCE = 5
DIRECTION = "+x,+y"
ROUNDS = 1
SOURCE_ROUNDS = 1
INITIAL_BASIS = "X"  # "X" or "Z"
```

```python
experiment = RotatedSurfaceEnlargeExperiment(
    distance=DISTANCE,
    target_distance=TARGET_DISTANCE,
    direction=DIRECTION,
    rounds=ROUNDS,
    source_rounds=SOURCE_ROUNDS,
    basis=INITIAL_BASIS,
)
circuit = experiment.build()
```
