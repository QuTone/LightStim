# QEC patch visualization

Set the distance below, then run the cells. `patch.visualize()` displays the code interactively; the final cell saves the same view as a standalone HTML file. This API supports self-contained `QECPatch` objects with at most 500 registered qubits (data and auxiliary combined). Coupler patches need surrounding system geometry and are not supported by this patch-only viewer.

```python
import sys
from pathlib import Path

ROOT = next(p for p in (Path.cwd(), *Path.cwd().parents)
            if (p / "lightstim/ir/qec_patch.py").is_file())
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from lightstim.qec_code.surface_code.rotated import RotatedSurfaceCode

DISTANCE = 3
```

```python
patch = RotatedSurfaceCode(distance=DISTANCE)
view = patch.visualize()
view
```

```python
output = view.save_html(ROOT / "build/examples/patch_visualization.html")
print(output.relative_to(ROOT))
```
