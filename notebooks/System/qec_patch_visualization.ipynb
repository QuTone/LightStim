# QEC structure visualization

Set the distance below and run the cells. The first view opens on all stabilizers of a rotated surface code. Select a check to inspect its signed Pauli support. The second view places two patches and a coupler on the same global coordinates. Each view accepts up to 500 registered qubits.

```python
import sys
from pathlib import Path

ROOT = next(p for p in (Path.cwd(), *Path.cwd().parents)
            if (p / "lightstim/ir/qec_patch.py").is_file())
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from lightstim.qec_code.surface_code.rotated import RotatedSurfaceCode

DISTANCE = 3
```

```python
patch = RotatedSurfaceCode(distance=DISTANCE)
view = patch.visualize(title=f"Rotated surface code · d = {DISTANCE}")
view
```

```python
output = view.save_html(ROOT / "build/examples/patch_visualization.html")
print(output.relative_to(ROOT))
```

## Two patches with an active XX seam

Select a patch or coupler name below the diagram to highlight its qubits. Search for `seam` in the operator list to see checks whose support spans both patches.

```python
from lightstim.ir.qec_system import QECSystem
from lightstim.qec_code.surface_code.rotated import RotatedTwoPatchCoupler

system = QECSystem()
system.add_patch(patch, name="left")
system.add_patch(patch, name="right", offset=(2 * DISTANCE + 2, 0))
system.register_coupler(RotatedTwoPatchCoupler(), ["left", "right"], "seam", interaction_type="XX")
system.activate_coupler("seam")
system.visualize(title="Two patches with an active XX seam")
```
